### c. Compute and sketch the empirical CDF for this sample.

The empirical cumulative distribution function (ECDF) is defined as:

$$
F_n(x) = \frac{1}{n}\sum_{i=1}^{n} I(X_i \leq x)
$$

It represents the proportion of observations that are less than or equal to a given value.

| x | Count <= x | F(x) |
|---|---|---|
| 1 | 1 | 0.2 |
| 3 | 3 | 0.6 |
| 6 | 4 | 0.8 |
| 11 | 5 | 1.0 |

The ECDF is:

$$
F_n(x) =
\begin{cases}
0, & x < 1 \\
0.2, & 1 \leq x < 3 \\
0.6, & 3 \leq x < 6 \\
0.8, & 6 \leq x < 11 \\
1, & x \geq 11
\end{cases}
$$


```python
X = np.array([1,3,3,6,11])
n = len(X)
ecdf = np.arange(1, n + 1) / n
plt.step(X, ecdf, where='post')
plt.xlabel('x')
plt.ylabel('F(x)')
plt.title('Empirical CDF (ECDF)')
plt.xticks([1, 3, 6, 11])
plt.yticks(np.arange(0, 1.1, 0.2))
plt.grid(True, alpha=0.3)
```

We often want to take transformations of random variables, particularly when doing feature engineering.

### a. Suppose we have a random variable X and transform it as Y = a + bX, with b > 0. The distribution of X is given by F_X(x). What is the distribution of Y, F_Y(y)? What is the density, f_Y(y)?

The distribution of Y is:

$$
F_Y(y) = P(Y \leq y)
$$

Substituting the transformation:

$$
F_Y(y) = P(a+bX \leq y)
$$

Since b > 0, we can rearrange the inequality:

$$
F_Y(y) = P\left(X \leq \frac{y-a}{b}\right)
$$

Therefore:

$$
\boxed{
F_Y(y) = F_X\left(\frac{y-a}{b}\right)
}
$$

The density of Y is obtained by differentiating the CDF:

$$
f_Y(y) = \frac{d}{dy}F_Y(y)
$$

Using the chain rule:

$$
\boxed{
f_Y(y) = \frac{1}{b}f_X\left(\frac{y-a}{b}\right)
}
$$

Suppose your sample is:

## Q3.

The exponential distribution is:

$$
F(x) =
\begin{cases}
0, & x < 0 \\
1-e^{-x}, & x \geq 0
\end{cases}
$$

### a. Sketch a graph of the exponential distribution and density.

The cumulative distribution function (CDF) is:

$$
F(x) =
\begin{cases}
0, & x < 0 \\
1-e^{-x}, & x \geq 0
\end{cases}
$$

The probability density function (PDF) is the derivative of the CDF:

$$
f(x) = \frac{d}{dx}F(x)
$$

Therefore:

$$
f(x) =
\begin{cases}
0, & x < 0 \\
e^{-x}, & x \geq 0
\end{cases}
$$

At x = 0:

$$
F(0)=0, \qquad f(0)=1
$$

Some values of the CDF and PDF are:

| x | F(x) | f(x) |
|---|---|---|
| 0 | 0.0000 | 1.0000 |
| 1 | 0.6321 | 0.3679 |
| 2 | 0.8647 | 0.1353 |
| 3 | 0.9502 | 0.0498 |
| 4 | 0.9817 | 0.0183 |
| 5 | 0.9933 | 0.0067 |

The CDF starts at 0 and increases toward 1. The PDF starts at 1 and decreases toward 0.





```python

x = np.array([0, 1, 2, 3, 4, 5])

F = np.array([0.0000, 0.6321, 0.8647, 0.9502, 0.9817, 0.9933])
f = np.array([1.0000, 0.3679, 0.1353, 0.0498, 0.0183, 0.0067])


fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# CDF plot
axes[0].plot(x, F, marker='o', color='blue')
axes[0].set_title('Exponential CDF')
axes[0].set_xlabel('x')
axes[0].set_ylabel('F(x)')
axes[0].set_ylim(0, 1.1)
axes[0].grid(True, alpha=0.3)

# PDF plot
axes[1].plot(x, f, marker='o', color='red')
axes[1].set_title('Exponential PDF')
axes[1].set_xlabel('x')
axes[1].set_ylabel('f(x)')
axes[1].set_ylim(0, 1.1)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### c. Suppose we transform X by Y = 1 + 3X. What is the distribution of Y? Provide a formula and sketch a graph in comparison to F_X. What is the probability that Y is less than 10?

The CDF of Y is:

$$
F_Y(y)=P(Y \leq y)
$$

Substituting the transformation:

$$
F_Y(y)=P(1+3X \leq y)
$$

Rearranging:

$$
F_Y(y)=P\left(X \leq \frac{y-1}{3}\right)
$$

Therefore:

$$
F_Y(y)=F_X\left(\frac{y-1}{3}\right)
$$

Substituting the exponential CDF:

$$
\boxed{
F_Y(y)=
\begin{cases}
0, & y<1 \\
1-e^{-(y-1)/3}, & y\geq1
\end{cases}
}
$$

The density is:

$$
\boxed{
f_Y(y)=
\begin{cases}
0, & y<1 \\
\frac{1}{3}e^{-(y-1)/3}, & y\geq1
\end{cases}
}
$$

To calculate the probability that Y is less than 10:

$$
P(Y<10)=P(1+3X<10)
$$

$$
=P\left(X<\frac{9}{3}\right)
$$

$$
=P(X<3)
$$

$$
=1-e^{-3}
$$

$$
\boxed{P(Y<10)=0.9502}
$$



```python



x = np.linspace(0, 15, 500)

FX = 1 - np.exp(-x)

# Transformed variable Y = 1 + 3X
y = 1 + 3*x
FY = 1 - np.exp(-(y - 1)/3)

# Plot both distributions
plt.figure(figsize=(10, 6))

plt.plot(x, FX, label='Original CDF: F_X(x)', color='blue')
plt.plot(y, FY, label='Transformed CDF: F_Y(y)', color='red')

# Mark y = 10 and its CDF value
plt.axvline(10, color='green', linestyle='--',
            label='y = 10')

plt.axhline(1 - np.exp(-3), color='orange',
            linestyle=':', label='F_Y(10) = 0.9502')

plt.xlabel('Value')
plt.ylabel('Cumulative Probability')
plt.title('Original vs. Transformed Exponential CDF')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```


## Q4.

The logistic distribution is:

$$
F(x)=\frac{1}{1+e^{-x}}
$$

### a. Sketch a graph of the logistic distribution and density.

The CDF is:

$$
F(x)=\frac{1}{1+e^{-x}}
$$

The PDF is obtained by differentiating the CDF:

$$
f(x)=\frac{d}{dx}\frac{1}{1+e^{-x}}
$$

Using the chain rule:

$$
\boxed{
f(x)=\frac{e^{-x}}{(1+e^{-x})^2}
}
$$

The density can also be written as:

$$
f(x)=F(x)(1-F(x))
$$

The logistic distribution is symmetric around zero:

$$
F(-x)=1-F(x)
$$

$$
f(-x)=f(x)
$$

At x = 0:

$$
F(0)=0.5
$$

$$
f(0)=0.25
$$

For the standard logistic distribution:

$$
\text{Mean}=\text{Median}=\text{Mode}=0
$$

$$
\text{Variance}=\frac{\pi^2}{3}
$$

The general logistic distribution has:

$$
F(x)=\frac{1}{1+e^{-(x-\mu)/\sigma}}
$$

Its density is:

$$
f(x)=
\frac{e^{-(x-\mu)/\sigma}}
{\sigma(1+e^{-(x-\mu)/\sigma})^2}
$$

where μ is the location parameter and σ > 0 is the scale parameter.

```python

# Generate x values
x = np.linspace(-6, 6, 100)

# Logistic CDF
F = 1 / (1 + np.exp(-x))
#pdf
f = np.exp(-x) / (1 + np.exp(-x))**2

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# CDF plot
axes[0].plot(x, F, color='blue')
axes[0].set_title('Logistic CDF')
axes[0].set_xlabel('x')
axes[0].set_ylabel('F(x)')
axes[0].grid(True, alpha=0.3)

# Pdf plot
axes[1].plot(x, f, color='red')
axes[1].set_title('Logistic PDF')
axes[1].set_xlabel('x')
axes[1].set_ylabel('f(x)')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### c. Suppose we transform X by Y = 2X - 1. What is the distribution of Y? Provide a formula and sketch a graph. What is the probability that Y is less than 0?

```python
y = np.linspace(-10, 10, 500)

# Transformed logistic CDF: Y = 2X - 1
FY = 1 / (1 + np.exp(-(y + 1) / 2))

# Transformed logistic PDF
fy = np.exp(-(y + 1) / 2) / (
    2 * (1 + np.exp(-(y + 1) / 2))**2
)

# Create two plots
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# CDF plot
axes[0].plot(y, FY, color='blue', label='Transformed CDF')
axes[0].axvline(0, color='green', linestyle='--',
                label='Y = 0')
axes[0].axhline(0.6225, color='orange', linestyle=':',
                label='P(Y < 0) = 0.6225')
axes[0].set_title('CDF of Y = 2X - 1')
axes[0].set_xlabel('y')
axes[0].set_ylabel('F_Y(y)')
axes[0].legend()
axes[0].grid(alpha=0.3)

# PDF plot
axes[1].plot(y, fy, color='red', label='Transformed PDF')
axes[1].set_title('PDF of Y = 2X - 1')
axes[1].set_xlabel('y')
axes[1].set_ylabel('f_Y(y)')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()


```

The median is the value x for which the probability that X is above or below x is 0.5:

$$
F(\text{median})=\frac{1}{2}
$$

### a. What is the median of the exponential distribution?

The exponential CDF with rate parameter λ is:

$$
F_X(x)=1-e^{-\lambda x}
$$

Set the CDF equal to 1/2:

$$
\frac{1}{2}=1-e^{-\lambda x}
$$

Rearranging:

$$
e^{-\lambda x}=\frac{1}{2}
$$

Taking the natural logarithm:

$$
-\lambda x=\ln\left(\frac{1}{2}\right)
$$

Therefore:

$$
\boxed{\text{Median}=\frac{\ln 2}{\lambda}}
$$

### b. What is the median of the logistic distribution?


The quantile function is the inverse of the distribution function. The CDF answers the question, "What fraction u of the time is X below x ?" ( F ( x ) = u ) and the quantile function answers the question, "For what value x is X below x with probability u ?" ( F − 1 ( u ) = x )

The logistic CDF is:

$$
F(x)=\frac{1}{1+e^{-(x-\mu)/\sigma}}
$$

Set the CDF equal to 1/2:

$$
\frac{1}{2}=
\frac{1}{1+e^{-(x-\mu)/\sigma}}
$$

Rearranging:

$$
1+e^{-(x-\mu)/\sigma}=2
$$

$$
e^{-(x-\mu)/\sigma}=1
$$

Taking the natural logarithm:

$$
-\frac{x-\mu}{\sigma}=0
$$

Therefore:

$$
\boxed{\text{Median}=\mu}
$$



### c. What is the quantile function of the exponential distribution?

The exponential CDF is:

$$
F(x)=1-e^{-\lambda x}
$$

Set F(x) = u:

$$
u=1-e^{-\lambda x}
$$

Rearranging:

$$
e^{-\lambda x}=1-u
$$

Taking the natural logarithm:

$$
-\lambda x=\ln(1-u)
$$

Therefore:

$$
\boxed{
F^{-1}(u)=-\frac{\ln(1-u)}{\lambda}
}
$$

for 0 < u < 1.

### d. What is the quantile function of the logistic distribution?

The logistic CDF is:

$$
F(x)=\frac{1}{1+e^{-(x-\mu)/\sigma}}
$$

Set F(x) = u:

$$
u=\frac{1}{1+e^{-(x-\mu)/\sigma}}
$$

Rearranging:

$$
1+e^{-(x-\mu)/\sigma}=\frac{1}{u}
$$

$$
e^{-(x-\mu)/\sigma}=\frac{1-u}{u}
$$

Taking the natural logarithm:

$$
-\frac{x-\mu}{\sigma}
=\ln\left(\frac{1-u}{u}\right)
$$

Therefore:

$$
\boxed{
F^{-1}(u)=
\mu+\sigma\ln\left(\frac{u}{1-u}\right)
}
$$

for 0 < u < 1.
